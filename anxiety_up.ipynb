```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Load dataset
data = pd.read_csv('/content/drive/MyDrive/new_bangladesh_student/new_Anxiety.csv')

# Columns for One-Hot Encoding
onehot_cols = [
    'Age',
    'Gender',
    'Academic Year',
    'Current CGPA'
]

# Columns for Label Encoding
label_cols = [
    'University',
    'Department'
]

# One-Hot Encoding (0/1 format)
data = pd.get_dummies(data, columns=onehot_cols, dtype=int)

# Label Encoding for University and Department
label_encoder = LabelEncoder()

for col in label_cols:
    data[col] = label_encoder.fit_transform(data[col])

# Custom encoding for Anxiety Level
anxiety_mapping = {
    'Minimal Anxiety': 0,
    'Mild Anxiety': 1,
    'Moderate Anxiety': 2,
    'Severe Anxiety': 3
}

data['Anxiety Level'] = data['Anxiety Level'].map(anxiety_mapping)

# Save processed dataset
save_path = '/content/drive/MyDrive/new_bangladesh_student/new_Anxiety1.csv'
data.to_csv(save_path, index=False)

print("Encoding completed successfully!")
print("Saved at:", save_path)

# Check encoded Anxiety Level distribution
print(data['Anxiety Level'].value_counts().sort_index())

data.head()
```
